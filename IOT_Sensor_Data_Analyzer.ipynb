```python

# IoT Sensor Data Analyzer

readings = [
    {"temperature": 25.5, "humidity": 45},
    {"temperature": 28.0, "humidity": 52},
    {"temperature": 35.0, "humidity": 70},
    {"temperature": 24.0, "humidity": 40},
    {"temperature": 19.0, "humidity": 35}
]

total_temp = 0
total_humidity = 0

print("===== IoT SENSOR DATA ANALYZER =====")

for i, reading in enumerate(readings, start=1):
    temp = reading["temperature"]
    humidity = reading["humidity"]

    total_temp += temp
    total_humidity += humidity

    print(f"\nReading {i}")
    print(f"Temperature: {temp} C")
    print(f"Humidity: {humidity}%")

    if temp > 30:
        print("Alert: High temperature!")
    elif temp < 18:
        print("Alert: Low temperature!")
    else:
        print("Temperature: Within demo limits")

    if humidity > 60:
        print("Alert: High humidity!")
    elif humidity < 30:
        print("Alert: Low humidity!")
    else:
        print("Humidity: Within demo limits")

print("\n===== SUMMARY REPORT =====")
print(f"Average temperature: {total_temp / len(readings):.2f} C")
print(f"Average humidity: {total_humidity / len(readings):.2f}%")
print(f"Total readings: {len(readings)}")
```

```python

# IoT Sensor Data Analyzer

readings = [
    {"temperature": 25.5, "humidity": 45},
    {"temperature": 28.0, "humidity": 52},
    {"temperature": 35.0, "humidity": 70},
    {"temperature": 24.0, "humidity": 40},
    {"temperature": 19.0, "humidity": 35}
]

total_temp = 0
total_humidity = 0

print("===== IoT SENSOR DATA ANALYZER =====")

for i, reading in enumerate(readings, start=1):
    temp = reading["temperature"]
    humidity = reading["humidity"]

    total_temp += temp
    total_humidity += humidity

    print(f"\nReading {i}")
    print(f"Temperature: {temp} C")
    print(f"Humidity: {humidity}%")

    if temp > 30:
        print("Alert: High temperature!")
    elif temp < 18:
        print("Alert: Low temperature!")
    else:
        print("Temperature: Within demo limits")

    if humidity > 60:
        print("Alert: High humidity!")
    elif humidity < 30:
        print("Alert: Low humidity!")
    else:
        print("Humidity: Within demo limits")

print("\n===== SUMMARY REPORT =====")
print(f"Average temperature: {total_temp / len(readings):.2f} C")
print(f"Average humidity: {total_humidity / len(readings):.2f}%")
print(f"Total readings: {len(readings)}")
```

```python

import matplotlib.pyplot as plt

temperatures = [r["temperature"] for r in readings]
humidities = [r["humidity"] for r in readings]
reading_numbers = range(1, len(readings) + 1)

plt.figure(figsize=(8, 4))
plt.plot(reading_numbers, temperatures, marker="o", label="Temperature (°C)")
plt.axhline(30, linestyle="--", label="High temperature limit")

plt.title("IoT Temperature Monitoring")
plt.xlabel("Reading Number")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.grid(True)
plt.show()
```

```python

plt.figure(figsize=(8, 4))

plt.plot(
    range(1, len(humidities) + 1),
    humidities,
    marker="o",
    label="Humidity (%)"
)

plt.axhline(
    y=60,
    linestyle="--",
    label="High humidity limit"
)

plt.title("IoT Humidity Monitoring")
plt.xlabel("Reading Number")
plt.ylabel("Humidity (%)")
plt.legend()
plt.grid(True)
plt.show()
```

```python

print("===== IoT SENSOR STATUS REPORT =====")

for i, reading in enumerate(readings, start=1):
    temp = reading["temperature"]
    humidity = reading["humidity"]

    if temp > 30:
        temp_status = "HIGH"
    elif temp < 18:
        temp_status = "LOW"
    else:
        temp_status = "NORMAL"

    if humidity > 60:
        humidity_status = "HIGH"
    elif humidity < 30:
        humidity_status = "LOW"
    else:
        humidity_status = "NORMAL"

    print(f"\nReading {i}")
    print(f"Temperature Status: {temp_status}")
    print(f"Humidity Status: {humidity_status}")

print("\n===== REPORT COMPLETE =====")
```

```python

temperatures = [r["temperature"] for r in readings]
humidities = [r["humidity"] for r in readings]

print("===== SENSOR ANALYSIS =====")

print("Highest temperature:", max(temperatures), "°C")
print("Lowest temperature:", min(temperatures), "°C")

print("Highest humidity:", max(humidities), "%")
print("Lowest humidity:", min(humidities), "%")

print("Temperature readings above 30°C:",
      sum(t > 30 for t in temperatures))

print("Humidity readings above 60%:",
      sum(h > 60 for h in humidities))
```

```python

import csv

filename = "iot_sensor_readings.csv"

with open(filename, "w", newline="") as file:
    writer = csv.DictWriter(
        file,
        fieldnames=["temperature", "humidity"]
    )

    writer.writeheader()
    writer.writerows(readings)

print("Sensor data saved successfully!")
print("File name:", filename)
print("Total records saved:", len(readings))
```
